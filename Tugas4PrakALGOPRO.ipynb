```python
belanjaan = ["Apel", "Susu", "Roti"]

harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

karakter_pertama = belanjaan[0][0]     
karakter_terakhir = belanjaan[-1][-1]   
print(f"Karakter pertama dari barang pertama: {karakter_pertama}")
print(f"Karakter terakhir dari barang terakhir: {karakter_terakhir}")
print("\n" + "="*45 + "\n")


total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]


total_belanja = total_apel + total_susu + total_roti

nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"

garis_ganda = "=" * 38
garis_putus = "-" * 38

struk = f"""{garis_ganda}
        {nama_toko}
{garis_ganda}
Pembeli : {nama_pembeli}

Barang\t\tHarga\tJumlah\tTotal
{belanjaan[0]}\t\t{harga_jumlah[0]}\t{harga_jumlah[1]}\t{total_apel}
{belanjaan[1]}\t\t{harga_jumlah[2]}\t{harga_jumlah[3]}\t{total_susu}
{belanjaan[2]}\t\t{harga_jumlah[4]}\t{harga_jumlah[5]}\t{total_roti}
{garis_putus}
Total Belanja\t\t\t{total_belanja}
{garis_ganda}
    Terima kasih sudah belanja! 🙏
{garis_ganda}"""

print(struk)
print("\n" + "="*45 + "\n")

belanjaan.append("Pisang")

print("Daftar belanjaan setelah penambahan barang baru:")
print(belanjaan)
```
